## Section 2 — Load & First Look at the Data

### 📥 How to get the data
1. Go to [https://www.kaggle.com/c/home-credit-default-risk/data](https://www.kaggle.com/c/home-credit-default-risk/data)
2. Accept the competition rules
3. Download **`application_train.csv`** and **`application_test.csv`**
4. Place them in a `data/raw/` folder next to this notebook

### 🗂️ What is this dataset?
Home Credit is a financial company that provides loans to people with little or no credit history.  
They collected **307,511 real loan applications** and recorded:
- **Demographics**: age, gender, education, family status
- **Financial info**: income, credit amount, annuity
- **External scores**: credit scores from external agencies
- **Target**: Did the client eventually default? (1 = YES, 0 = NO)


````python
# Load the training dataset
# The training set has both features AND the target (1 = defaulted, 0 = repaid)
df = pd.read_csv("data/raw/application_train.csv")

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
````

````python
# Preview the first 5 rows
df.head()
````

````python
# Get a concise summary — column names, data types, non-null counts
df.info(verbose=False, show_counts=True)
````

````python
# Statistical summary of numerical columns
df.describe().T.style.background_gradient(cmap="Blues")
````

### 🎯 The Target Variable

The column `TARGET` is what we want to predict:
- `0` → client **repaid** the loan (no default)  
- `1` → client **defaulted** on the loan

This is a **binary classification** problem.  
Think of it as: *"Given everything we know about this person, will they pay back the loan?"*


````python
# Count of each class
target_counts = df["TARGET"].value_counts()
print(target_counts)
print(f"
Default rate: {target_counts[1] / len(df) * 100:.2f}%")
````

````python
# Visualize the class distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Bar chart
target_counts.plot(kind="bar", ax=axes[0], color=["steelblue", "tomato"], edgecolor="black")
axes[0].set_title("Loan Outcome Counts")
axes[0].set_xlabel("Target (0 = Repaid, 1 = Defaulted)")
axes[0].set_ylabel("Number of Applications")
axes[0].set_xticklabels(["Repaid (0)", "Defaulted (1)"], rotation=0)

# Pie chart
axes[1].pie(target_counts, labels=["Repaid", "Defaulted"],
            autopct="%1.1f%%", colors=["steelblue", "tomato"],
            startangle=90, explode=(0, 0.05))
axes[1].set_title("Class Distribution (%)")

plt.suptitle("Class Imbalance — Most Clients Repay Their Loans", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("reports/figures/class_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
````

### ⚠️ Notice the Class Imbalance!

About **91.9%** of clients repaid → only **8.1%** defaulted.

💡 **Why this matters:** If we train a naive model that always predicts "repaid", it would be  
91.9% accurate — but it would **miss every single defaulter**! That's useless for a bank.

We'll deal with this later using special techniques (SMOTE, class weights).  
For now, just keep this imbalance in mind as you look at every chart.
